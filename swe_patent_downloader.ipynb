```python
import os
import csv
import time
import requests
import urllib3

# Suppress warnings that appear when SSL verification is disabled
urllib3.disable_warnings(urllib3.exceptions.InsecureRequestWarning)

# --- Configuration ---
# Update this to the exact name of your CSV file
OUTPUT_DIR = '../data/swe_patent_1895_1900'
CSV_FILE_PATH = '../data/swepat_url_1895_1900.csv' 

# 1. Changed to http:// instead of https://
BASE_URL = 'http://swepat.iprhistory.org/document/pdf/'

def download_pdfs():
    os.makedirs(OUTPUT_DIR, exist_ok=True)

    if not os.path.exists(CSV_FILE_PATH):
        print(f"Error: Could not find CSV file at '{CSV_FILE_PATH}'")
        return

    with open(CSV_FILE_PATH, mode='r', encoding='utf-8') as file:
        reader = csv.reader(file)
        
        header = next(reader, None)
        print(f"Found headers: {header}")
        print("-" * 30)

        for row_num, row in enumerate(reader, start=2):
            if not row or len(row) < 2:
                continue 
            
            patent_number = row[0].strip()
            document_id = row[1].strip()

            url = f"{BASE_URL}{document_id}"
            output_filepath = os.path.join(OUTPUT_DIR, f"{patent_number}.pdf")

            if os.path.exists(output_filepath):
                print(f"Skipping {patent_number}.pdf (Already exists)")
                continue

            print(f"Downloading Patent {patent_number}...")
            
            try:
                # 2. Added verify=False to bypass strict SSL/TLS handshake checks
                response = requests.get(url, stream=True, timeout=15, verify=False)
                
                if response.status_code == 200:
                    with open(output_filepath, 'wb') as pdf_file:
                        for chunk in response.iter_content(chunk_size=8192):
                            if chunk:
                                pdf_file.write(chunk)
                    print(f" -> Saved to {output_filepath}")
                else:
                    print(f" -> Failed. HTTP Status: {response.status_code} for URL: {url}")
            
            except requests.RequestException as e:
                print(f" -> Network error downloading {patent_number}: {e}")
            
            time.sleep(1)

if __name__ == "__main__":
    download_pdfs()
```
